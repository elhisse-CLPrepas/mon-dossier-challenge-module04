## 1. Ouvrez la source

Dans le guide de la séance 12, l’étape 5 demande de conserver V1 et V2 et de comparer leur contenu. Le journal présente le passage avant et après, avec le motif de la correction. Lisez aussi les deux repères dans `fiche-source-s12.md`. Demandez qui pourra décider de l’usage du message.

```python
titre = "Produire avec l’IA sous responsabilité humaine — Guide pratique"
statut_source = "V2 — À RELIRE / NON VALIDÉE"
fait_guide = "La V2 demande de conserver les deux versions et de comparer leur contenu."
fait_journal = "Le journal présente les corrections avant et après et en donne les raisons."
action = "Souhaitez-vous voir un extrait du journal de corrections ?"
print(titre)
print(statut_source)
print("G1 ·", fait_guide)
print("J1 ·", fait_journal)
```

## 2. Observez ce que Python assemble

Le formateur choisit le public et le canal. Le gabarit, lui, se contente d’insérer des valeurs. Les brouillons A et B sont ensuite lus à voix haute. Quel passage de la fiche soutient chaque affirmation ?

```python
message_gabarit = f"« {titre} ». {fait_guide} {action}"
print("Gabarit Python")
print(message_gabarit)

ouverture = "Vous relisez un support produit avec l’IA."
promesse = "Ce guide garantit un document sans erreur."
brouillons = {
    "A": f"{ouverture} {fait_journal} {action}",
    "B": f"{ouverture} {promesse} {action}",
}
for nom, texte in brouillons.items():
    print(f"\nBrouillon {nom} · simulation")
    print(texte)
```

## 3. Prédisez le résultat d’une règle

Nous choisissons 400 caractères comme seuil pour cet exercice. Python compte les espaces et la ponctuation. Avant d’exécuter, demandez au groupe quelle indication s’affichera à 400 et à 401 caractères. La recherche de la phrase d’action compte les occurrences exactes ; c’est au lecteur de repérer un éventuel deuxième appel formulé autrement.

```python
def verifier_longueur(texte, seuil=400):
    return "À raccourcir" if len(texte) > seuil else "Longueur acceptable"

for nom, texte in brouillons.items():
    print(f"{nom} · {len(texte)} caractères | {verifier_longueur(texte)} | "
          f"action écrite repérée {texte.count(action)} fois")

for taille in (400, 401):
    print(f"{taille} caractères : {verifier_longueur('x' * taille)}")
```

## 4. Corrigez le contenu

Marquez la promesse de B, revenez à G1 et J1, puis proposez un remplacement. Après la discussion, exécutez la cellule pour afficher l’arbitrage préparé par le formateur. La fonction de remplacement applique le choix humain déjà inscrit.

```python
correction_choisie = fait_journal
message_corrige = brouillons["B"].replace(promesse, correction_choisie)
print("Avant :", promesse)
print("Après :", correction_choisie)
print("Motif : le journal conserve les corrections. Il ne promet aucun résultat.")
print("Brouillon B corrigé :", message_corrige)
print("Statut du message : BROUILLON — À RELIRE")
```
